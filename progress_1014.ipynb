```python
# 2. standard = close, SMA, EMA, RSI, MACD, bollinger bands, ATR

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# 1) 데이터 불러오기
data = yf.download("AAPL", start="2024-01-01", end="2025-01-01")

```

```python
data
```

```python
data.columns = data.columns.droplevel('Ticker')
```

```python
data
```

```python
import talib

data["SMA20"] = talib.SMA(data["Close"], timeperiod=20)
data["EMA60"] = talib.EMA(data["Close"], timeperiod=60)
data["SMA20"] = talib.RSI(data["Close"], timeperiod=14)
data["MACD"], data["MACD_SIGNAL"], data["MACD_HIST"] = talib.MACD(data["Close"])
upper, middle, lower = talib.BBANDS(data["Close"], timeperiod=20)
data["BB_upper"], data["BB_middle"], data["BB_lower"] = upper, middle, lower
data["ATR"] = talib.ATR(data["High"], data["Low"], data["Close"], timeperiod=14)
```

```python
data
```

```python
# 1. make y(label) as column -> return -> 0,1 binary
# (tomorrow_price-today_price)/today_price ->

data['Daily Return'] = data["Close"].pct_change()
data.tail()
```

```python
data.loc[data['Daily Return']>0,'Return'] = 1
data.loc[data['Daily Return']<=0,'Return'] = 0
```

```python
data.tail()
```

```python
import seaborn as sns
import pandas as pd

# 독립변수(X): 수치형 feature 선택
X = data[["SMA20", "EMA60", "MACD", "ATR", "Daily Return"]]

# 종속변수(y): 
y = data["Return"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from datetime import date
```

```python
# ticker = st.sidebar.text_input("Ticker: ")
ticker = "AAPL"
START = "2024-01-01"
TODAY = date.today().strftime("%Y-%m-%d")


def load_data(ticker):
    data = yf.download(ticker, START, TODAY)
    data.reset_index(inplace=True)
    return data

# st.sidebar.text("Loading stock data...")
data = load_data(ticker)
# st.sidebar.success("Data loaded successfully!")

# Data Analysis Section
data.columns = data.columns.droplevel('Ticker')

# Daily Return Calculation: 
data['Daily Return'] = data['Close'].pct_change()
data['Daily Return'].plot(figsize=(10,4), title=f"{ticker} Daily Return")
plt.show()

data["SMA20"] = talib.SMA(data["Close"], timeperiod=20)
data["EMA60"] = talib.EMA(data["Close"], timeperiod=60)
data["SMA20"] = talib.RSI(data["Close"], timeperiod=14)
data["MACD"], data["MACD_SIGNAL"], data["MACD_HIST"] = talib.MACD(data["Close"])
upper, middle, lower = talib.BBANDS(data["Close"], timeperiod=20)
data["BB_upper"], data["BB_middle"], data["BB_lower"] = upper, middle, lower
data["ATR"] = talib.ATR(data["High"], data["Low"], data["Close"], timeperiod=14)
```
